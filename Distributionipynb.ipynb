```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Define lambda and create a Poisson distribution object
lam = 2
pois_dist = poisson(mu=lam)

# Generate 100 random samples from the Poisson distribution
samples = pois_dist.rvs(size=100)

# Plot the PMF and CDF of the distribution
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# PMF
x = np.arange(0, 10)
pmf = pois_dist.pmf(x)
ax[0].bar(x, pmf)
ax[0].set_xlabel('Number of events')
ax[0].set_ylabel('Probability')
ax[0].set_title(f'Poisson PMF (lambda={lam})')

# CDF
cdf = pois_dist.cdf(x)
ax[1].step(x, cdf)
ax[1].set_xlabel('Number of events')
ax[1].set_ylabel('Cumulative probability')
ax[1].set_title(f'Poisson CDF (lambda={lam})')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson, norm

# Define lambda and create a Poisson distribution object
lam = 2
pois_dist = poisson(mu=lam)

# Generate population data from the Poisson distribution
population = pois_dist.rvs(size=100000)

# Take 10,000 samples of size 50 from the population
sample_means = np.zeros(10000)
for i in range(10000):
    sample = np.random.choice(population, size=50, replace=False)
    sample_means[i] = np.mean(sample)

# Plot the sample means and compare to the normal distribution
fig, ax = plt.subplots(1, 1, figsize=(8, 4))

# Histogram of sample means
ax.hist(sample_means, bins=30, density=True, alpha=0.5)

# Overlay normal distribution with mean and variance estimated from samples
mean_est = np.mean(sample_means)
var_est = np.var(sample_means, ddof=1)
x = np.linspace(mean_est - 4*np.sqrt(var_est), mean_est + 4*np.sqrt(var_est), 100)
y = norm.pdf(x, loc=mean_est, scale=np.sqrt(var_est))
ax.plot(x, y, color='r', linewidth=2)

# Set plot title and labels
ax.set_title(f'Sample Means from Poisson Distribution (lambda={lam})')
ax.set_xlabel('Sample Mean')
ax.set_ylabel('Probability Density')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import poisson, norm

lam = 10  # Poisson distribution parameter
n = 10000  # Number of samples to generate
sample_sizes = [2, 5, 10, 30]  # Different sample sizes to try

fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.ravel()  # Flatten the 2D array of subplots

for i, size in enumerate(sample_sizes):
    means = []
    for _ in range(n):
        samples = poisson.rvs(mu=lam, size=size)
        means.append(np.mean(samples))
    sns.histplot(data=means, bins=30, ax=axs[i], kde=True)
    axs[i].set_title(f'Sample size: {size}')
    axs[i].set_xlabel('Sample mean')
    axs[i].set_ylabel('Density')

    # Plot vertical lines to highlight mean and standard deviation
    mean = np.mean(means)
    std = np.std(means)
    x_vals = np.linspace(mean-4*std, mean+4*std, 100)
    y_vals = norm.pdf(x_vals, mean, std)
    axs[i].plot(x_vals, y_vals, color='red', linestyle='--')
    axs[i].axvline(x=mean, color='green', linestyle='--', label=f'Mean: {mean:.2f}')
    axs[i].axvline(x=mean+std, color='orange', linestyle='--', label=f'Standard deviation: {std:.2f}')
    axs[i].axvline(x=mean-std, color='orange', linestyle='--')

fig.suptitle('Central Limit Theorem using a Poisson distribution')
plt.tight_layout()
plt.show()
```

### Convergence in probability

```python
import numpy as np
import matplotlib.pyplot as plt

# Set up the parameters
alpha = 5
beta = 2
mu = 30
sigma = 2
n = 10000

# Generate the data
X = np.random.normal(mu, sigma, n)
epsilon = np.random.normal(0, 4, n)
Y = alpha + beta * X + epsilon

# Compute the sample means for different sample sizes
sample_sizes = np.arange(10, n + 1, 10)
sample_means = []
for size in sample_sizes:
    sample = np.random.choice(Y, size=size, replace=False)
    sample_means.append(np.mean(sample))
print("Sample mean:", sample_means[-1])

# Plot the results
plt.plot(sample_sizes, sample_means, 'o-', label='Sample mean')
plt.axhline(y=np.mean(Y), color='r', linestyle='-', label='Population mean')
plt.xlabel('Sample size')
plt.ylabel('Sample mean')
plt.title('Convergence in Probability')
plt.legend()
plt.show()
```
