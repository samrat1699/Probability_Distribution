```python
import random

def bernoulli(p, n):
    """
    Generate n random samples from a Bernoulli distribution with success probability p.
    """
    samples = []
    for i in range(n):
        if random.random() < p:
            samples.append(1)
        else:
            samples.append(0)
    return samples

# Example usage
p = 0.7
n = 10
samples = bernoulli(p, n)

print(samples)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

# Define the probability of success
p = 0.6

# Create a Bernoulli distribution object
dist = bernoulli(p)

# Generate a random sample of size n from the distribution
n = 10000
sample = dist.rvs(size=n)

# Calculate the sample means for different sample sizes
sample_means = []
sample_sizes = [10, 50, 100, 500, 1000, 5000]   # sample sizes to use
for size in sample_sizes:
    means = []
    for i in range(1000):
        s = np.random.choice(sample, size=size, replace=True)
        means.append(np.mean(s))
    sample_means.append(means)

# Plot the distribution of sample means for different sample sizes
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(10, 6))
axes = axes.flatten()
for i, ax in enumerate(axes):
    ax.hist(sample_means[i], bins=30, alpha=0.5, color='blue')
    ax.set_xlabel('Sample mean')
    ax.set_ylabel('Frequency')
    ax.set_title(f'Sample size: {sample_sizes[i]}')
    ax.axvline(x=p, color='r', linestyle='-', label='Population mean')
    ax.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

def binomial_pmf(n, p):
    """Calculate the probability mass function (PMF) of a binomial distribution."""
    pmf = []
    for k in range(n+1):
        coef = 1
        for i in range(k):
            coef *= (n-i)/(i+1)
        pmf.append(coef * p**k * (1-p)**(n-k))
    return pmf

n = 10   # number of trials
p = 0.5  # probability of success

# Calculate the PMF
pmf = binomial_pmf(n, p)

# Plot the PMF
plt.bar(range(n+1), pmf)
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution (n={}, p={})".format(n, p))
plt.show()
```

```python
import math
import random
import matplotlib.pyplot as plt

# Define the number of trials and probability of success
n = 10
p = 0.5

# Generate a list of random variables X
X = [sum([random.random() < p for j in range(n)]) for i in range(10000)]

# Calculate the mean and variance of X
mean = sum(X) / len(X)
variance = sum([(x - mean)**2 for x in X]) / len(X)

# Calculate the PMF for X
PMF = [sum([1 for x in X if x == k]) / len(X) for k in range(n+1)]

# Plot the histogram of X
plt.hist(X, bins=n+1, density=True, align='left', rwidth=0.8)
plt.xticks(range(n+1))
plt.xlabel('Number of successes')
plt.ylabel('Probability')
plt.title('Binomial Distribution (n={}, p={})'.format(n, p))

# Plot the PMF for X
plt.plot(range(n+1), PMF, 'ro', markersize=3)

# Print the mean, variance, and PMF for X
print('Mean:', mean)
print('Variance:', variance)
print('PMF:', PMF)

plt.show()
```

```python
import math

def binomial_pmf(n, p, k):
    """
    Calculates the probability mass function of a binomial distribution with parameters n and p,
    evaluated at k, using the formula P(X=k) = (n choose k) * p^k * (1-p)^(n-k).
    """
    if k < 0 or k > n:
        return 0
    return math.comb(n, k) * p**k * (1-p)**(n-k)

# Example usage
n = 10
p = 0.5
for k in range(n+1):
    print(f'P(X={k}) = {binomial_pmf(n, p, k)}')
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

def binomial_pmf(n, p, k):
    """Calculate the probability mass function of a binomial distribution."""
    coef = math.comb(n, k)
    prob = coef * p**k * (1-p)**(n-k)
    return prob

# Define the parameters of the binomial distribution
n = 50
p = 0.4

# Calculate the mean of the binomial distribution
mean = n * p
print(f"Mean: {mean:.4f}")

# Calculate the median of the binomial distribution
median = np.floor(mean)
if (n*p - median) < 0.5:
    median += 1
print(f"Median: {median}")

# Calculate the mode of the binomial distribution
mode = np.round((n+1) * p - 1e-10)
print(f"Mode: {mode}")

# Calculate the skewness of the binomial distribution
skewness = (1 - 2*p) / np.sqrt(n*p*(1-p))
print(f"Skewness: {skewness:.4f}")

# Calculate the PMF for each possible value of k
k_values = np.arange(0, n+1)
pmf_values = [binomial_pmf(n, p, k) for k in k_values]

# Plot the PMF as a line chart
plt.plot(k_values, pmf_values)

# Label the plot with the mean, median, mode, and skewness
plt.axvline(mean, color='r', linestyle='--', label=f"Mean: {mean:.4f}")
plt.axvline(median, color='g', linestyle='--', label=f"Median: {median}")
plt.axvline(mode, color='b', linestyle='--', label=f"Mode: {mode}")
plt.title(f"Binomial distribution (n={n}, p={p}, skewness={skewness:.4f})")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.legend()

# Show the plot
plt.show()
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

def binomial_pmf(n, p, k):
    """Calculate the probability mass function of a binomial distribution."""
    coef = math.comb(n, k)
    prob = coef * p**k * (1-p)**(n-k)
    return prob

# Define the parameters of the binomial distribution
n = 100
p = 0.3

# Calculate the mean and variance of the binomial distribution
mu = n * p
sigma = np.sqrt(n * p * (1 - p))
print(f"Binomial: mu={mu:.4f}, sigma={sigma:.4f}")

# Calculate the PMF for each possible value of k
k_values = np.arange(0, n+1)
pmf_values = [binomial_pmf(n, p, k) for k in k_values]

# Plot the PMF as a line chart
plt.plot(k_values, pmf_values)

# Plot the normal distribution that approximates the binomial distribution
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
pdf_values = 1/(sigma * np.sqrt(2*np.pi)) * np.exp(-(x - mu)**2 / (2*sigma**2))
plt.plot(x, pdf_values, color='r')

# Label the plot
plt.title(f"Binomial vs. Normal distribution (n={n}, p={p})")
plt.xlabel("Number of successes")
plt.ylabel("Probability density")
plt.legend(["Binomial", "Normal"])

# Show the plot
plt.show()
```

## Poisson

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Define the parameter lambda of the Poisson distribution
lam = 2.5

# Calculate the PMF for each possible value of k in the Poisson distribution
k_values = np.arange(0, 11)
pmf_values = [math.exp(-lam) * lam**k / math.factorial(k) for k in k_values]

# Plot the PMF of the Poisson distribution as a histogram
plt.hist(k_values, bins=np.arange(0, 12)-0.5, weights=pmf_values, density=True)

# Label the plot and show the legend
plt.title(f"Poisson distribution (lambda={lam})")
plt.xlabel("Number of events")
plt.ylabel("Probability density")

# Show the plot
plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Define the parameter lambda of the Poisson distribution
lam = 2.5

# Define the range of values for k
k_values = np.arange(0, 11)

# Calculate the PMF for each value of k
pmf = [math.pow(lam, k) * math.exp(-lam) / math.factorial(k) for k in k_values]

# Plot the PMF as a bar chart
plt.bar(k_values, pmf)

# Label the axes and title
plt.xlabel('k')
plt.ylabel('P(X = k)')
plt.title(f'Poisson distribution with lambda = {lam}')

# Calculate the mean, median, and mode
mean = poisson.mean(lam)
median = poisson.median(lam)
mode = math.floor(lam)

# Calculate the skewness and kurtosis
skewness = poisson.stats(lam, moments='s')
kurtosis = poisson.stats(lam, moments='k')

# Print the mean, median, mode, skewness, and kurtosis
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Skewness: {skewness}")
print(f"Kurtosis: {kurtosis}")

# Display the plot
plt.show()
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# Define the parameter lambda of the Poisson distribution
lam = 2.5

# Define the range of values for k
k_values = np.arange(0, 11)

# Calculate the PMF for each value of k
pmf = [math.pow(lam, k) * math.exp(-lam) / math.factorial(k) for k in k_values]

# Calculate the mean, median, mode, skewness, and kurtosis
mean = lam
median = math.floor(lam + 1/3 - 0.02/lam)
mode = math.floor(lam)
skewness = skew(pmf)
kurtosis = kurtosis(pmf)

# Plot the PMF as a bar chart
plt.bar(k_values, pmf)

# Label the axes and title
plt.xlabel('k')
plt.ylabel('P(X = k)')
plt.title(f'Poisson distribution with lambda = {lam}')

# Print the mean, median, mode, skewness, and kurtosis
print(f"Mean: {mean:.2f}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Skewness: {skewness:.2f}")
print(f"Kurtosis: {kurtosis:.2f}")

# Display the plot
plt.show()
```
