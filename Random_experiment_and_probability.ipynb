suppose that a fair coin is tossed three times. compute the probability of obtaining (a) exactly one head, and (b) a least one tail

P(X = k) = (n choose k) * p^k * (1 - p)^(n - k)


---
where n is the number of trials, k is the number of successes, p is the probability of success (which is 0.5 for a fair coin), and (n choose k) is the binomial coefficient.


```python
import math

# A list of possible outcomes
outcomes = ['H', "T"]
sample_space = []

for outcome1 in outcomes:
  for outcome2 in outcomes:
    for outcome3 in outcomes:
      sample_space.append(outcome1 + outcome2 + outcome3)

print(sample_space)


# Compute the probability of obtaining exactly one head
p_one_head = (math.comb(3, 1) * (1/2)**1 * (1/2)**2)
print("Probability of obtaining exactly one head:", p_one_head)


# Compute the probability of obtaining at least one tail
p_at_least_one_tail = 1 - (1/2)**3
print("Probability of obtaining at least one tail:", p_at_least_one_tail)
```

## card experiment

```python
suits = ["club", "spade", "heart", "diamond"]
cards = ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"]
jokers = ["Joker1", "Joker2"]
pack = []

# Create the pack of cards without jokers
for suit in suits:
    for card in cards:
        pack.append(card + " of " + suit)

# Add the jokers to the pack
pack += jokers

# Print the sample space
print(pack)
```
